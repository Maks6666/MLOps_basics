```python
import pandas as pd
import numpy as np 
from sklearn.preprocessing import OneHotEncoder
```

```python
link = '/Users/maxkucher/preprocessing/mlops/income/data.csv'
data = pd.read_csv(link)
data
```

```python
data["income"].unique()
data['income'] = data['income'].replace({'<=50K': 0, '>50K': 1})
data
```

```python
data.isnull().sum()
```

```python
# data.loc[2, "workclass"]
```

```python
data['workclass'].unique()
data['workclass'] = data['workclass'].replace({'?': 'Unkown'})
data
```

```python
encoder = OneHotEncoder(handle_unknown = 'ignore', sparse_output = False)

encoded = encoder.fit_transform(data[['workclass']])

encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['workclass']), index=data.index)

data = pd.concat([data.drop('workclass', axis="columns"), encoded_df], axis='columns')

data
```

```python
data = data.drop('fnlwgt', axis='columns')
data
```

```python
data['education'].unique()
```

```python
value = 'No-HS'
data['education'] = data['education'].replace({'7th-8th': value, '10th': value, '1st-4th': value, '5th-6th': value, '9th': value, '11th': value, '12th': value})
data['education'].unique()
```

```python
data
```

```python
data['occupation'] = data['occupation'].replace({'?': np.nan})
data
```

```python
mode = data['occupation'].mode()[0]
data['occupation'] = data['occupation'].fillna(mode)
data
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoded = encoder.fit_transform(data[['education']])
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['education']), index=data.index)
data = pd.concat([data.drop('education', axis='columns'), encoded_df], axis='columns')
data
```

```python
data['marital.status'].unique()
```

```python
value = 'Married'
data['marital.status'] = data['marital.status'].replace({'Married-civ-spouse': value, 'Married-spouse-absent': value, 'Married-AF-spouse': value})
data
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

encoded = encoder.fit_transform(data[['marital.status']])

endoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['marital.status']), index=data.index)

data = pd.concat([data.drop('marital.status', axis='columns'), endoded_df], axis='columns')

data
```

```python
data['race'].unique()

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

encoded = encoder.fit_transform(data[['race']])

encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['race']), index=data.index)

data = pd.concat([data.drop('race', axis='columns'), encoded_df], axis='columns')

data
```

```python
data['sex'] = data['sex'].replace({'Female': 0, 'Male': 1})
data
```

```python
data['native.country'].unique()
```

```python
data['native.country'] = data['native.country'].replace({'?': np.nan})
data['native.country'].isnull().sum()
```

```python
data = data.dropna(subset='native.country')
data['native.country'].isnull().sum()
```

```python
data
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False, min_frequency=50)
encoded = encoder.fit_transform(data[['native.country']])

encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['native.country']), index=data.index)
data = pd.concat([data.drop('native.country', axis='columns'), encoded_df], axis='columns')

data
```
