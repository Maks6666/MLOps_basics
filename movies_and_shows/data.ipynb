```python
import pandas as pd
import numpy as np 
```

```python
link = "/Users/maxkucher/preprocessing/mlops/movies_and_shows/data.csv"
data = pd.read_csv(link)
data
```

```python
data.isnull().sum()
```

```python
data["director"] = data["director"].fillna("Unknown")
data["cast"] = data["cast"].fillna("Unknown")
data["country"] = data["country"].fillna("Unknown")
```

```python
data.isnull().sum()
```

```python
data = data.dropna(subset=["date_added"])
data = data.dropna(subset=["rating"])
data = data.dropna(subset=["duration"])
```

```python
data.isnull().sum()
```

```python
data
```

```python
data["type"].unique()
```

```python
data["type"] = data["type"].replace({"Movie": 0, "TV Show": 1})
```

```python
data
```

```python
data["country"] = data["country"].str.split(",").str[0].str.strip()
data
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False, min_frequency=20)
encoded = encoder.fit_transform(data[["country"]])
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["country"]), index=data.index)

data = pd.concat([data.drop("country", axis="columns"), encoded_df], axis="columns")
data
```

```python
data["date_added"] = pd.to_datetime(data["date_added"], errors='coerce')
data
```

```python
data['year'] = data["date_added"].dt.year
data['month'] = data["date_added"].dt.month
data = data.drop("date_added", axis="columns")
data
```

```python
# !IMPORTANT - Cycle endocing 

data["month_sin"] = np.sin(2 * np.pi * data["month"] / 12)
data["month_cos"] = np.cos(2 * np.pi * data["month"] / 12)
data = data.drop('month', axis="columns")
data
```

```python
data["rating"].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False, min_frequency=20)

encoded = encoder.fit_transform(data[["rating"]])
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["rating"]), index=data.index)
data = pd.concat([data, encoded_df], axis="columns")
data
```

```python
def parse_duration(value: str):
    if pd.isna(value):
        return np.nan, np.nan
    
    value = value.lower()
    
    if "min" in value:
        minutes = int(value.replace("min", "").strip())
        return minutes, 0

    if "season" in value:
        seasons = int(value.replace("seasons", "").replace("season", "").strip())
        return 0, seasons
    
    return np.nan, np.nan


data[["duration_minutes", "duration_seasons"]] = (data["duration"].apply(parse_duration).apply(pd.Series))
data = data.drop("duration", axis="columns")
        
```

```python
data = data.drop(["show_id", "title", "director", "cast"], axis="columns")
data
```

```python
data["listed_in"].unique()
```

```python
data["genres"] = data["listed_in"].str.split(',').apply(lambda x: [g.strip() for g in x])
data = data.drop("listed_in", axis="columns")
data
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
genres_encoded = mlb.fit_transform(data["genres"])
genres_df = pd.DataFrame(genres_encoded, columns=[f"column_{g}" for g in mlb.classes_], index=data.index)
data = pd.concat([data.drop("genres", axis="columns"), genres_df], axis="columns")
data
```
