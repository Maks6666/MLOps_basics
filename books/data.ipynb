```python
columns_to_drop = ['index', 'Book Name', 'Author']
data = data.drop(columns_to_drop, axis='columns')
data
```

```python
data['language_code'].value_counts()
data['language_code'] = data['language_code'].replace({'en-US': 'eng', 'en-GB': 'eng', 'en-CA': 'eng'})
data['language_code'].value_counts()
```

```python
ara = data[data['language_code'] == 'ara']['is_successful']
eng = data[data['language_code'] == 'eng']['is_successful']
fre = data[data['language_code'] == 'fre']['is_successful']
nl = data[data['language_code'] == 'nl']['is_successful']
spa = data[data['language_code'] == 'spa']['is_successful']

rates = (ara.mean(), eng.mean(), fre.mean(), nl.mean(), spa.mean())
plt.bar(['ara', 'eng', 'fre', 'nl', 'spa'], rates)
```

```python
data = data.drop('language_code', axis='columns')
data
```

```python
def plot_bar(data, column, metric):
    grouped = data.groupby(column)[metric].mean()

    values = grouped.index
    rates = grouped.values

    plt.bar(values, rates)
    plt.title(f'% of sucess among {column}')
    plt.show()
```

```python
data['genre'] = data['genre'].replace({"genre fiction": "fiction"})
```

```python
columns = ['genre', 'Author_Rating', 'Publisher ']

for column in columns:
    plot_bar(data, column, 'is_successful')
```

```python
data['Publishing Year'].value_counts()
```

```python
def year(x):
    if x < 1980:
        return 'old'
    elif x < 2000:
        return 'classic'
    elif x < 2010:
        return 'modern'
    else:
        return 'recent'
    

data['year_group'] = data['Publishing Year'].apply(year)
data
```

```python
plot_bar(data, 'year_group', 'is_successful')
```

```python
columns_to_drop = ['gross sales', 'publisher revenue', 'sales rank', 'units sold']
data = data.drop(columns_to_drop, axis='columns')
data
```

```python
def rating(x):
    if x < 3:
        return 'low'
    elif x < 4:
        return 'medium'
    else:
        return 'high'
    

data['rating_cat'] = data['Book_average_rating'].apply(rating)

data
```

```python
def plot_hist(data, column, bins):
    plt.hist(data[column], bins=bins)
    plt.title(f'Distribution of {column}')
    plt.show()
```

```python
column = 'Book_ratings_count'
plot_hist(data, column, 30)
```

```python
# columns = ['Book_ratings_count', 'Publishing Year', 'sale price']
columns = {'Book_ratings_count': 30, 'sale price': 8}

for key, value in columns.items():
    plot_hist(data, key, value)
```
